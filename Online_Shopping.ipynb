```python
import time
import pandas as pd
from tabulate import tabulate

# Initializing
global df
df=pd.read_csv("Products.csv") #Catalog Dataframe
global cart_df  #Cart Dataframe 

# Function to Clear the Screen
def clearScreen():
    from IPython.display import clear_output
    clear_output(wait=True)
    pass

# Function for displaying login screen
def login_screen():
    print('''
    ********************************************************
    ********************************************************
                  Welcome to Demo Marketplace
    ********************************************************
    ********************************************************
    ''')
    print('Enter 1 to Login as User')
    print('Enter 2 to Login as Admin')
    print('Enter 0 to quit')
    choice=int(input('\n Enter Choice :'))
    if choice==1:
        l_user="User"
    elif choice==2:
        l_user="Admin"
    return choice,l_user

# Function for displaying Main Menu
def main_menu():
    # print('''
    # ****************************************************
    #                     Main Menu 
    # ****************************************************
    # ''')
    print('Enter 3 to View Catalog')
    print('Enter 4 to Add Items to Cart')
    print('Enter 5 to View Items in Cart')
    print('Enter 6 to Remove Items to Cart')
    print('Enter 0 to quit')
    choice=int(input('\n Enter Choice :'))
    return choice

def admin_main_menu():
    # print('''
    # ****************************************************
    #                     Main Menu 
    # ****************************************************
    # ''')
    print('Enter 7 to View Catalog')
    print('Enter 8 to Add Items to Catalog')
    print('Enter 9 to Remove Items from Catalog')
    print('Enter 0 to quit')
    choice=int(input('\n Enter Choice :'))
    return choice

# Function to view catalog
def view_catalog():
    clearScreen()
    # txt='''
    # ***************************************************************
    #                        Product Catalog 
    # ***************************************************************
    #     '''
    hd=['Category Id','Category Name','Product Id','Product Name','Available In store','Unit Price']
    print(tabulate(df, headers=hd, tablefmt='fancy_grid'))
    return df

# Function to add items to cart
def add_to_cart():
    response ='Y'
    global cart
    cart={'product_id':[],'quantity':[]}
    while (response.upper() != "N"):
        try:
            prd_id=int(input("Enter Product ID :"))
            qty=int(input("Enter Quantity:"))
            #print((df.loc[df.product_id==prd_id,'available_in_store']).item())
            if (df.loc[df.product_id==prd_id,'available_in_store']).item()>=qty:
                print("Item Available")
                cart['product_id'].append(prd_id)
                cart['quantity'].append(qty)
            else:
                print("Item Unavailable")
            #print(cart)
        except:
            print("Oops ! Something went wrong!. Please check your order again.")
        response=input("Continue shopping?(Y/N)")
    return(cart)

def add_items_to_catalog():
    cat_id=input("Enter category Id:")
    cat_nm=input("Enter category Name:")
    prd_id=int(input("Enter Product Id:"))
    prd_nm=input("Enter Product Name:")
    qty=int(input("Enter quantity available in store:"))
    prc=input("Enter unit price:")
    df.loc[len(df.index)]=[cat_id,cat_nm,prd_id,prd_nm,qty,prc]


# Function to reduce items from catalog
def reduce_from_catalog(product_id,quantity):
    pid=product_id
    qty=quantity  
    df.loc[df.product_id==pid,'available_in_store']=df.loc[df.product_id==pid,'available_in_store']-qty
    
     #df['product_id']==101
    #filtered_df = df.query('product_id == 101')
    #df.loc[0,'available_in_store']=12

#Function for Payment
def Payment():
    print("Choose Payment Method")
    print("Enter 7 for PAYPAL")
    print("Enter 8 for Credit/Debit")
    print("Enter 0 to exit Payment")
    ch=int(input("Enter Choice :"))
    if ch==7 or ch==8:
        print("Thanks for the payment!")
        
        cart.clear()
    elif ch==0:
        print("Your order is successfully placed.Thanks for the payment!")
        print("exiting.. Goodbye!")
        return 
   
#Function to process user/admin choices              
def process_choice(choice):
    if choice==1:
        print("\n Welcome User !!")
    elif choice==2:
        print("\n Welcome Admin !!")
    elif choice==3 or choice==7:
        print("View Catalog")
        print("============")
        view_catalog()
    elif choice==4:
        global cart_df
        print("Add Items to cart")
        print("=================")
        add_to_cart()
        cart_df=pd.DataFrame(cart)
    elif choice==5: 
        cart_df
        print("View Items in cart")
        print(tabulate(cart_df, headers=["Product Id","Quantity"], tablefmt='fancy_grid'))
        response=input("Proceed to Payment?(Y/N)")
        if (response.upper()=="Y"):
            Payment()    
    elif choice==6:
        print("Remove Items from cart")
        ch="Y"
        cart_df
        cart_df=pd.DataFrame(cart)
        while ch.upper()=="Y":
            prd_id=int(input("Enter Product Id that needs to be removed from cart:"))                 
            cart_df=cart_df.loc[cart_df.product_id!=prd_id]
            ch=input("Do you want to remove more items?(Y/N):")
            if ch.upper()=="N":
                             print("returning to main menu..")
                             break
    elif choice==8:
        print("Add items to catalog")
        add_items_to_catalog()
    elif choice==9:
        print("Remove Items from catalog")
        ch="Y"
        global df
        while ch.upper()=="Y":
            prd_id=int(input("Enter Product Id that needs to be removed from catalog:"))                 
            df=df.loc[df.product_id!=prd_id]
            ch=input("Do you want to remove more items?(Y/N):")
            if ch.upper()=="N":
                             print("returning to main menu..")
                             return
                        
    elif choice==0:
        print("exiting.. Goodbye!")
        return
    else:
        print("Please enter valid input")

    # except:
    #         print("Something went wrong!")
        

#Main Program

choice,l_user=login_screen()
process_choice(choice)      
time.sleep(1)       
clearScreen()
while choice !=0:
    if l_user=="User":
        choice=main_menu()
        process_choice(choice)
    elif l_user=="Admin":
        choice=admin_main_menu()
        process_choice(choice)
```
